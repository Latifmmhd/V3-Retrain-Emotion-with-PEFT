```python
import os
import gdown
import pandas as pd
from datasets import load_dataset, Dataset, DatasetDict
from sklearn.model_selection import train_test_split
```

```python
file_id = "1oyriEp5xWkL2efEfrZw-R1qC_tOHNS2Y"
url = f"https://drive.google.com/uc?id={file_id}"

# Download the file
output = "dataset/all_data.csv"
gdown.download(url, output, quiet=False)

# Load the dataset
df = pd.read_csv(output)
print(df.columns.tolist())

# Kolom yang dipakai: tweet (teks) dan label (emosi)
df = df.dropna(subset=['tweet', 'label'])
df['label'] = df['label'].str.strip()

print(df.shape)
print(df['label'].value_counts())
df.head()
```

```python
# Split into train and temp (which will be further split into val and test)
train_df, temp_df = train_test_split(df, test_size=0.3, random_state=42)

# Split temp into validation and test sets
val_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42)
```

```python
# Save each DataFrame to CSV
train_df.to_csv('dataset/train.csv', index=False)
val_df.to_csv('dataset/val.csv', index=False)
test_df.to_csv('dataset/test.csv', index=False)
```
